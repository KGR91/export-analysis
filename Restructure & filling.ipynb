```python
# 1. Deleted first 4 rows in a source file

import pandas as pd
import os
from datetime import datetime

# path to file
file_path = r"C:\Users\Георгий\Downloads\09 month_2026_country_goods.xlsx"

# read file without first 4 lines

df = pd.read_excel(file_path, sheet_name="Аркуш1", skiprows=4, header=None)

# clean colomn names from empty spaces
df.columns = df.columns.astype(str).str.strip()

# save a new file
date_str = datetime.now().strftime("%Y%m%d_%H%M%S")
output_file = os.path.join(
    os.path.dirname(file_path),
    f"trimmed_country_goods_{date_str}.xlsx"
)

df.to_excel(output_file, index=False)

print(f"✅ File is ready:\n{output_file}")
```

```python
# 2. Filled empty cells in a column "0" (country name) to make tab structure more SQL-friendly.

import pandas as pd

# path to file
file_path = r"C:\Users\Георгий\Downloads\trimmed_country_goods_20261025_153756.xlsx"

# read a necessary sheet
df = pd.read_excel(file_path, sheet_name="Sheet1")

# fill empty cells in a column "0" with a data above.
df["0"] = df["0"].ffill()

# save into the same file by changing sheet
with pd.ExcelWriter(file_path, engine="openpyxl", mode="a", if_sheet_exists="replace") as writer:
    df.to_excel(writer, sheet_name="Sheet1", index=False)
```
